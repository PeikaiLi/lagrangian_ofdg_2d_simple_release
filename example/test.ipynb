```julia
include("../src/load_all.jl")
```

```julia
case_shu = ShuOsherCase.make_case(
    nx = 100,
    ny = 2,
    degree = 2,
    OF = true,
    characteristic = true,
    positivity = false,
    cfl = 0.1,
    max_retries = 4,
)

solution_shu = Solver.raw_solve_2d(
    case_shu;
    time_method = :SSPRK3,
    verbose = true,
    save_every = 20,
    dt_max = case_shu.final_time / 20,
)
# 16mins
```

```julia
println("total retry = $(solution_shu.number_of_retries)")
display(Visualization.plot_mesh(solution_shu.state,case_shu.mesh))
Diagnostics.conservation_report(solution_shu)

for field in (:density, :pressure, :velocity_x, :velocity_y)
    gif_result = Visualization.draw_gif(
        solution_shu,
        "../results/case_shu$(field).gif";
        field = field,
        fps = 10,
    )
    display(gif_result)
end
# plot this is very very slow!!!!!!!!!!!
```
